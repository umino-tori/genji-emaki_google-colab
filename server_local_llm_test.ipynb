```python
#「server_local_llm.ipynb」で表示された「🚀 URL:https://xxxxxxxxxxxxxxxx..ngrok-free.app」を記載してください
BASE_URL =  "https://afff-abc-eff-77-77.ngrok-free.app"
```

```python
import requests
# POSTリクエスト（JSONボディ）
res = requests.post(f"{BASE_URL}/llms", json={
    'system_prompt': "やあ、調子はどう？",
    'max_tokens': 512,
    'temperature':0.7,
    'top_p': 0.9
})
print(res.json()['response'])
```
